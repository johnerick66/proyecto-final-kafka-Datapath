```python
import json
import time
from kafka import KafkaProducer
```

```python
# 1. Configuración del Producer
# Cambia 'localhost:9092' por 'kafka:9092' si estás dentro de la red Docker.
bootstrap_servers = 'localhost:29092'
topic_name = 'proyect-final-topic'

producer = KafkaProducer(
    bootstrap_servers=bootstrap_servers,
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

print(f"Conectado a Kafka en {bootstrap_servers}. Enviando mensajes...")
```

```python

# 2. Envío de 5 mensajes simples
mensajes = [
    {"id": 1, "usuario": "Ana", "accion": "login", "timestamp": time.time()},
    {"id": 2, "usuario": "Carlos", "accion": "compra", "timestamp": time.time()},
    {"id": 3, "usuario": "Beatriz", "accion": "click_banner", "timestamp": time.time()},
    {"id": 4, "usuario": "David", "accion": "logout", "timestamp": time.time()},
    {"id": 5, "usuario": "Elena", "accion": "registro", "timestamp": time.time()}
]

for msg in mensajes:
    # Envío asíncrono
    future = producer.send(topic_name, value=msg)
    # Esperar confirmación
    record_metadata = future.get(timeout=10)
    print(f"Mensaje enviado con éxito: {msg} -> Partición: {record_metadata.partition}, Offset: {record_metadata.offset}")
    time.sleep(1) # Pausa de 1 segundo entre envíos

# Asegurar que todos los mensajes en búfer se envíen
producer.flush()
producer.close()
print("Proceso de envío finalizado.")

```
