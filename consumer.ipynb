```python
import json
from kafka import KafkaConsumer
```

```python
# 1. Configuración del Consumer
bootstrap_servers = 'localhost:29092'
topic_name = 'proyect-final-topic'

consumer = KafkaConsumer(
    topic_name,
    bootstrap_servers=bootstrap_servers,
    auto_offset_reset='earliest',      # Lee desde el principio del topic
    enable_auto_commit=True,
    group_id='grupo-actividad-1',       # Nombre del grupo de consumidores
    value_deserializer=lambda m: json.loads(m.decode('utf-8'))
)

print(f"Escuchando mensajes en el topic '{topic_name}'...\n")
```

```python
# 2. Lectura y verificación de los mensajes
contador = 0
try:
    for message in consumer:
        contador += 1
        print(f"[Mensaje {contador}] Recibido desde Partición {message.partition}, Offset {message.offset}:")
        print(f" Contenido: {message.value}\n")
        
        # Detener al recibir los 5 mensajes esperados
        if contador == 5:
            print("Se han recibido y verificado los 5 mensajes con éxito.")
            break
except KeyboardInterrupt:
    print("Consumidor detenido manualmente.")
finally:
    consumer.close()
```
